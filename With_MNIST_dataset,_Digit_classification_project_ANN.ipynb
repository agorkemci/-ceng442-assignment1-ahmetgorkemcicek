```python
def get_data_loader(batch_size=64):
  #her iterasyonda işlenecek veri miktarı=batch size
  #örneğin 64000lik sample verin var 64000/64=1000 iterasyon
  #1000 iterasyon sonunda veri train edilmiş olacak
  #amaç ise belleği şişirmemek
  transform=transforms.Compose([transforms.ToTensor(),transforms.Normalize(0.5,0.5)])
  #görüntüyü nparrayden => tensore(matrix) çevirir ve ölçeklendirir 0-255=> 0-1
  #pixel değerlerini de -1 ile 1  arasına ölçekler
  #mnist veri setini indir ve train test split
  train_set=torchvision.datasets.MNIST(root="./data",train=True, download=True,transform=transform)
  test_set=torchvision.datasets.MNIST(root="./data",train=False, download=True,transform=transform)
  #aynı dizinde data folder ı oluşturup zaten halihazırda bulunan mnisti koyuyoruz

  #pytorch veri yükleyicisi
  # Belleği şişirmemek ve batch'ler halinde işlemek için DataLoader oluşturuyoruz
  train_loader = torch.utils.data.DataLoader(
      train_set, batch_size=batch_size, shuffle=True
  )
  #verisetinde rakamlar sürekli artarda geliyor 00000000 1111111 gibi dolayısıyla genelleme yeteneğini artırmak için

  test_loader = torch.utils.data.DataLoader(
      test_set, batch_size=batch_size, shuffle=False
  )
  """
  train_set / test_set: Verinin kendisidir. Veriyi ve labelları çeker, RAM'e yükler. Tek başına indeksi bellidir; örneğin train_set[0] dediğinde bana 0. resmi ve etiketini verir. Ancak model bu şekilde bütün veriyi tek seferde yutamaz.

  DataLoader: Dataset üzerine giydirilen bir aktarım yöneticisidir.
  Belirlenen batch_size kadar veriyi arkada paketler, isterse karıştırır ve modele iterasyon iterasyon sunar."""




  return train_loader,test_loader


train_loader,test_loader=get_data_loader()




"""
Veriler $0$ ile $1$ arasında olduğunda, tüm değerler pozitif olur. Bu durum, arka arkaya gelen katmanlardaki nöronların girdilerinin hep aynı yönde (pozitif) olmasına yol açar.
Girdiler ortalama etrafında simetrik (örneğin $-1$ ile $+1$ arasında, ortalaması $0$) olduğunda, gradyanlar daha dengeli güncellenir ve ağın öğrenmesi daha hızlı ve kararlı ilerler.
Tanh gibi bazı aktivasyon fonksiyonlarının çıktı aralığı $[-1, 1]$'dir. Girdilerin de benzer bir aralıkta olması,
nöronların çok erken aşamada doygunluğa (gradient vanishing/saturation) ulaşıp öğrenmenin durmasını engeller."""
```
